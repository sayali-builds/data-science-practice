```python
import numpy as np
from scipy.stats import zscore
data = [1,2,3,4,11,250,1,12,13]   #list
arr = np.array(data)    #array
z_scores = zscore(arr)
print(z_scores)
threshold = 2
outliers = arr[np.abs(z_scores) > threshold]
print("Outliers in above data are:", outliers)
```

```python
import numpy as np

def detect_outliers(data, threshold=3):
    mean = np.mean(data)
    std = np.std(data)
    outliers = [x for x in data if abs((x - mean) / std) > threshold]
    return outliers

# Example usage:
data = [10, 12, 11, 13, 10, 11, 12, 11]
print("Outliers:", detect_outliers(data))
```

```python
def detect_outliers_iqr(data):
    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    return [x for x in data if x < lower_bound or x > upper_bound]

# Example usage:
print("Outliers (IQR):", detect_outliers_iqr(data))
```
